```python
import os
from pathlib import Path

def batch_replace_hexo_covers():
    # 你的目标文件夹路径 (使用 r 前缀防止转义字符报错)
    directory_path = r"E:\hexo_kratos_v3.0\source\_posts"
    
    # 需要替换的字符串
    old_string = "pic:"
    new_string = "cover:"
    
    # 将路径转换为 Path 对象
    folder = Path(directory_path)
    
    # 检查文件夹是否存在
    if not folder.exists() or not folder.is_dir():
        print(f"❌ 找不到文件夹：{directory_path}，请检查路径是否正确！")
        return

    modified_count = 0
    
    # 使用 rglob 遍历该文件夹及所有子文件夹下的所有 .md 文件
    # 如果你不想包含子文件夹，可以将 rglob 换成 glob
    for file_path in folder.rglob("*.md"):
        try:
            # 读取文件内容 (Hexo 文件一般是 utf-8 编码)
            with open(file_path, 'r', encoding='utf-8') as file:
                content = file.read()
            
            # 检查文件中是否包含需要替换的内容
            if old_string in content:
                # 执行替换
                new_content = content.replace(old_string, new_string)
                
                # 将修改后的内容写回文件
                with open(file_path, 'w', encoding='utf-8') as file:
                    file.write(new_content)
                
                print(f"✅ 已修改: {file_path.name}")
                modified_count += 1
                
        except Exception as e:
            print(f"❌ 处理文件 {file_path.name} 时出错: {e}")

    print("-" * 30)
    print(f"🎉 替换完成！共成功修改了 {modified_count} 个 Markdown 文件。")

if __name__ == "__main__":
    batch_replace_hexo_covers()
```

```python
import os
import re

# ===== 配置 =====
folder_path = r"E:\blog\source\_posts"  # 替换为你的目标文件夹路径

# 匹配 Markdown 图片语法：![alt](path)
md_image_pattern = re.compile(r'!\[([^\]]*)\]\(([^)]+)\)')
# 匹配 pic: xxx 形式（行内或行首），不区分大小写
# 这里尽量只匹配形如 "pic: something" 的部分，避免误伤别的单词
pic_pattern = re.compile(r'(?i)\bpic:\s*([^\s)]+)')

# ===== 遍历文件夹 =====
for filename in os.listdir(folder_path):
    if not filename.endswith(".md"):
        continue

    filepath = os.path.join(folder_path, filename)
    file_base = os.path.splitext(filename)[0]  # 去掉后缀的文件名

    with open(filepath, "r", encoding="utf-8") as f:
        content = f.read()

    # 用可变容器存放标记，避免 nonlocal/global 问题
    updated_flag = [False]

    # ===== 替换 Markdown 图片 =====
    def replace_md_image(match):
        alt_text = match.group(1)
        img_path = match.group(2)
        # 跳过已经是 /images/ 开头的路径
        if img_path.startswith("/images/"):
            return match.group(0)
        updated_flag[0] = True
        # 保持原有相对路径结构，只在前面加 /images/<file_base>/
        return f"![{alt_text}](/images/{file_base}/{img_path})"

    content = md_image_pattern.sub(replace_md_image, content)

    # ===== 替换 pic: xxx =====
    def replace_pic(match):
        img_path = match.group(1)
        # 跳过已经是 /images/ 开头的路径
        if img_path.startswith("/images/"):
            return match.group(0)
        updated_flag[0] = True
        return f"pic: /images/{file_base}/{img_path}"

    content = pic_pattern.sub(replace_pic, content)

    # ===== 写回文件 =====
    if updated_flag[0]:
        with open(filepath, "w", encoding="utf-8") as f:
            f.write(content)
        print(f"✅ 已更新：{filename}")
    else:
        print(f"— 无需修改：{filename}")
```
